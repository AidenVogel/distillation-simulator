```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 0.011, 0.022, 0.037, 0.059, 0.065, 0.071, 0.090, 0.105, 0.113, 0.134, 0.172, 0.199, 0.230, 0.275, 0.337,
     0.381, 0.412, 0.444, 0.469, 0.498, 0.516, 0.544, 0.572, 0.605, 0.629, 0.659, 0.689, 0.720, 0.767,
     0.782, 0.824, 0.849, 0.866, 0.874, 0.884, 0.894, 0.899, 0.910, 0.923, 0.942, 0.950, 0.967, 0.986, 1])
y = np.array([0, 0.097, 0.178, 0.259, 0.347, 0.366, 0.384, 0.432, 0.462, 0.475, 0.501, 0.533, 0.547, 0.563, 0.577, 0.598,
     0.611, 0.618, 0.631, 0.641, 0.655, 0.664, 0.671, 0.684, 0.701, 0.715, 0.731, 0.746, 0.764, 0.796,
     0.805, 0.839, 0.856, 0.871, 0.878, 0.886, 0.895, 0.899, 0.909, 0.921, 0.940, 0.948, 0.965, 0.984, 1])

xD = 0.85 # desired top-poroduct purity (mole fraction)
R = 3 # reflux ratio
xF = 0.20 # ethanol in the feed
xB = 0.02 # ethanol allowed in the bottom product

fig, ax = plt.subplots(figsize = (6,6))

ax.plot(x, y, marker = 'o', color = 'r', linestyle = '-', label = 'Ethanol-water equilibrium')
ax.plot([0, 1], [0, 1], color = 'b', linestyle = '-', label = 'y = x')

# rectifying (top-section) operating line
slope = R / (R + 1)
intercept = xD / (R + 1)
x_op = np.array([xF, xD])
y_op = slope * x_op + intercept
ax.plot(x_op, y_op, color = 'g', linestyle = '-', label = f'Operating Line (R = {R})')

# where the top line meets the feed line
y_meet = slope * xF + intercept

# feed line (verticle, bc feed is a liquid at boiling point)
ax.plot([xF, xF], [xF, y_meet], color = 'orange', linestyle = '-', label = 'Feed Line')

# bottom operating line
ax.plot([xB, xF], [xB, y_meet], color = 'm', linestyle = '-', label = 'Bottom Operating Line')
ax.plot(xB, xB, marker = 's', color = 'k', linestyle = '', label = 'Bottoms')

ax.plot(xD, xD, marker = 's', color = 'k', linestyle = '', label = 'Target Purity')

ax.set(xlabel = 'Ethanol in liquid (mole fraction)', ylabel = 'Ethanol in vapor (mole fraction)', title = 'Ethanol-Water Vapor-Liquid Equilibrium (1 atm)')
ax.legend()
ax.grid()
ax.set_xlim(0,1)
ax.set_ylim(0,1)
plt.show()
```
